```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Descargar datos históricos de BTC
btc = yf.download("BTC-USD", start="2020-01-01")

print(btc.shape)
btc.head(10)
```

```python
# Gráfico del precio de cierre histórico
plt.figure(figsize=(14, 6))
plt.plot(btc["Close"], color="orange", linewidth=1.5)
plt.title("Precio histórico de Bitcoin (BTC-USD)", fontsize=16)
plt.xlabel("Fecha")
plt.ylabel("Precio en USD")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("../images/precio_historico.png")  # guarda la imagen
plt.show()
```

```python
# Medias móviles SMA 50 y SMA 200
btc["SMA_50"] = btc["Close"].rolling(window=50).mean()
btc["SMA_200"] = btc["Close"].rolling(window=200).mean()

plt.figure(figsize=(14, 6))
plt.plot(btc["Close"], label="Precio BTC", color="orange", linewidth=1.2, alpha=0.8)
plt.plot(btc["SMA_50"], label="Media móvil 50 días", color="blue", linewidth=1.5)
plt.plot(btc["SMA_200"], label="Media móvil 200 días", color="red", linewidth=1.5)

plt.title("Bitcoin - Precio y Medias Móviles", fontsize=16)
plt.xlabel("Fecha")
plt.ylabel("Precio en USD")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
